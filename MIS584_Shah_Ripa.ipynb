1. read the file using the code covered in Section 5 (i.e., File I/O) of the Python tutorial






```python
# open the file
file = open("/content/drive/MyDrive/Colab Notebooks/Data/mis584_float.txt", "r")

# read all the text into a single string
all_content = file.read()
print(all_content)
file.close()
print()
```

2. store the numbers in the file into a Python list; how many numbers are there in the file?

```python
#initialize a new variable with empty list
number_list = []
#initialize a new variable with floating point number
number = 0.0
# 1. Use .splitlines() to separate the single string into a list of line strings
for line in all_content.splitlines():
  #convert each line to float
  try:
    number = float(line.strip())
    number_list.append(number)
  except ValueError:
    print(f"Skip non-numeric line:{line.strip()}")
print(number_list)
```

3. create a lambda function that returns 0 if the input number is negative; otherwise returns the square root of the input number; use the functional operators discussed in Section 6 (i.e., Python Functional Programming) to transform the list in Step 2 to a new list, such that all negative numbers are transformed to 0 and all the other numbers are transformed to their square root;

```python
import numpy as np
import math as math
#first method without using lamda function
new_number_list = []
for num in number_list:
    if num < 0:
      new_number_list.append(0)
    else:
      new_number_list.append(math.sqrt(num))
print(new_number_list)
```

```python
# Second method with using lamda function
new_number_list = list(map(lambda x: 0 if x < 0 else math.sqrt(x), number_list))
print(new_number_list)
```

```python
#Define the name of the output file
output_filename = "transformed_numbers.txt"


# export the converted string to another external file
out_file = open("/content/drive/MyDrive/Colab Notebooks/Data/transformed_numbers.txt", "w")
out_file.write(str(new_number_list))
out_file.close()
print(f"Successfully wrote {len(new_number_list)} transformed numbers to '{output_filename}'")
```
